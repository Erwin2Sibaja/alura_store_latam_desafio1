```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import FuncFormatter
```

##**Resultados de ventas por tienda**

```python
etiquetas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
ventas = [total_ventas_tienda, total_ventas_tienda2, total_ventas_tienda3, total_ventas_tienda4]

fig, ax = plt.subplots(figsize=(10, 6))

for data, label in zip(ventas, etiquetas):
    ax.bar(label, data, label=label)

ax.set_ylabel('Ventas')

ax.set_title('Distribución de ventas por tienda',
            fontsize=16, fontweight='bold')

def millones_formato(x, pos):
    return f'{x/1e6:.0f} millones $'

ax.yaxis.set_major_formatter(FuncFormatter(millones_formato))
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.pie(x=ventas, labels=etiquetas, autopct='%1.1f%%', startangle=90, explode=(0,0,0,0.1))

ax.set_title('Distribución de ventas por tienda',
            fontsize=16, fontweight='bold')

ax.axis('equal')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(etiquetas, ventas, color='green', marker='o', linestyle='-')

ax.set_title('Distribución de ventas por tienda',
            fontsize=16, fontweight='bold')
ax.yaxis.set_major_formatter(FuncFormatter(millones_formato))
plt.show()
```

##**Ventas de productos por categoría**

###**Combinar datos de top categoria de ventas**

```python
top_categorias_dict = {
    'Tienda 1': top_3_categorias_tienda,
    'Tienda 2': top_3_categorias_tienda2,
    'Tienda 3': top_3_categorias_tienda3,
    'Tienda 4': top_3_categorias_tienda4
}

top_categorias_df = pd.DataFrame(top_categorias_dict)
top_categorias_df.index.name = 'Categoría'
top_categorias_df
```

```python
fig, ax = plt.subplots(figsize=(12, 8))
categorias = top_categorias_df.index
tiendas = top_categorias_df.columns
bar_width = 0.15

rango = np.arange(len(categorias))

for i in range(len(tiendas)):
    ax.bar(rango + i * bar_width, top_categorias_df[tiendas[i]], width=bar_width, label=tiendas[i])

ax.set_xlabel('Categoría')
ax.set_ylabel('Ventas')
ax.set_title('Comparativa de Ventas de las 3 Categorías Principales por Tienda', fontsize=16, fontweight='bold')
ax.set_xticks(rango + bar_width)
ax.set_xticklabels(categorias)
ax.legend()

plt.show()
```

```python
from os import replace
categorias_dict = {}

for categoria in top_categorias_df.index:
    for tienda in top_categorias_df.columns:
        clave = f'{categoria.replace(" ", "_")}_{tienda.replace(" ", "")}'
        valor = [top_categorias_df.loc[categoria, tienda]]

        categorias_dict[clave] = valor

categorias_df = pd.DataFrame(categorias_dict)
categorias_df
```

```python
categorias = top_categorias_df.index
tiendas = top_categorias_df.columns

display(categorias, tiendas)
```

```python
num_categorias = len(categorias)
filas = int(np.ceil(np.sqrt(num_categorias)))
columnas = int(np.ceil(num_categorias / filas))

fig, axes = plt.subplots(filas, columnas, figsize=(14, 12))
axes = np.atleast_1d(axes).flatten()

for i, categoria in enumerate(categorias):
    ax = axes[i]
    ventas_categoria = top_categorias_df.loc[categoria]
    min_valor = ventas_categoria.min()
    explode = [0.1 if valor == min_valor else 0 for valor in ventas_categoria]
    ax.pie(ventas_categoria, labels=tiendas, autopct='%1.1f%%', startangle=90, explode=explode)
    ax.set_title(f'Ventas por Tienda: {categoria}', fontsize=14, fontweight='bold')
    ax.axis('equal')

fig.suptitle('Comparativa de Ventas de las 3 Categorías Principales por Tienda',
             fontsize=20,
             fontweight='bold',
             y=1.02)

for j in range(num_categorias, filas * columnas):
    fig.delaxes(axes[j])
```
