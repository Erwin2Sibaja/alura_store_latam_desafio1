```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler
from matplotlib import colormaps
from matplotlib.cm import ScalarMappable
```

##**Resultados de ventas por tienda**

```python
tiendas = ['Tienda 1', 'Tienda 2', 'Tienda 3', 'Tienda 4']
ventas = [total_ventas_tienda, total_ventas_tienda2, total_ventas_tienda3, total_ventas_tienda4]

fig, ax = plt.subplots(figsize=(10, 6))

ax.pie(x=ventas, labels=tiendas, autopct='%1.1f%%', startangle=90, explode=(0,0,0,0.1))

ax.set_title('Distribución de ventas por tienda',
            fontsize=16, fontweight='bold')

ax.axis('equal')

plt.show()
```

##**Resultados ventas de productos por categoría**

###**Combinar datos de top categoria de ventas**

```python
top_categorias_dict = {
    'Tienda 1': top_3_categorias_tienda,
    'Tienda 2': top_3_categorias_tienda2,
    'Tienda 3': top_3_categorias_tienda3,
    'Tienda 4': top_3_categorias_tienda4
}

top_categorias_df = pd.DataFrame(top_categorias_dict)
top_categorias_df.index.name = 'Categoría'
top_categorias_df
```

```python
fig, ax = plt.subplots(figsize=(12, 8))
categorias = top_categorias_df.index
bar_width = 0.15

rango = np.arange(len(categorias))

for i in range(len(tiendas)):
    ax.bar(rango + i * bar_width, top_categorias_df[tiendas[i]], width=bar_width, label=tiendas[i])

ax.set_xlabel('Categoría')
ax.set_ylabel('Ventas')
ax.set_title('Comparativa de Ventas de las 3 Categorías Principales por Tienda', fontsize=16, fontweight='bold')
ax.set_xticks(rango + bar_width)
ax.set_xticklabels(categorias)
ax.legend()

plt.show()
```

```python
productos_ventas = {
    'Tienda 1': {'Más vendido': producto_mas_vendido_tienda, 'Menos vendido': producto_menos_vendido_tienda},
    'Tienda 2': {'Más vendido': producto_mas_vendido_tienda2, 'Menos vendido': producto_menos_vendido_tienda2},
    'Tienda 3': {'Más vendido': producto_mas_vendido_tienda3, 'Menos vendido': producto_menos_vendido_tienda3},
    'Tienda 4': {'Más vendido': producto_mas_vendido_tienda4, 'Menos vendido': producto_menos_vendido_tienda4}
}

unidades_mas_vendido = []
for tienda in tiendas:
    producto_str = productos_ventas[tienda]['Más vendido']
    unidades = int(producto_str.split(': ')[1])
    unidades_mas_vendido.append(unidades)

df_plot = top_categorias_df.T.copy()

df_plot['P. Más Vendido'] = unidades_mas_vendido

df_plot_T = df_plot.T
metrics = df_plot_T.index

fig, ax = plt.subplots(figsize=(14, 8))
bar_width = 0.18
rango = np.arange(len(tiendas))

colores_metrics = plt.cm.Set2.colors[:len(metrics)]

for i, metric in enumerate(metrics):
    ax.bar(
        rango + i * bar_width,
        df_plot_T.loc[metric],
        width=bar_width,
        label=metric,
        color=colores_metrics[i],
        alpha=0.8
    )

ax.set_title('Comparativa de Magnitud de Unidades Vendidas (Categorías vs. Producto Estrella)', fontsize=16, fontweight='bold')
ax.set_xlabel('Tienda', fontsize=12)
ax.set_ylabel('Unidades Vendidas (Magnitud Absoluta)', fontsize=12)

ax.set_xticks(rango + (len(metrics)/2 - 0.5) * bar_width)
ax.set_xticklabels(tiendas)

ax.grid(axis='y', linestyle='--', alpha=0.6)
ax.legend(title='Métrica de Venta')

plt.tight_layout()
plt.show()
```

##**Ventas, Calificaciones y gastos**

```python
dict_datos_general = pd.DataFrame({
    'Tienda': list(tiendas),
    'Ventas': [total_ventas_tienda, total_ventas_tienda2, total_ventas_tienda3, total_ventas_tienda4],
    'Calificaciones': [promedio_satisfaccion_cliente, promedio_satisfaccion_cliente2, promedio_satisfaccion_cliente3, promedio_satisfaccion_cliente4],
    'Costo de envío promedio': [costo_de_envio, costo_de_envio2, costo_de_envio3, costo_de_envio4]
    })
```

```python
df_analisis = pd.DataFrame(dict_datos_general)
df_analisis.head()
```

```python
scaler = MinMaxScaler()

df_resultado = df_analisis.copy()
df_resultado[['Ventas', 'Calificaciones']] = scaler.fit_transform(df_resultado[['Ventas', 'Calificaciones']])
df_resultado['Costo de envío promedio'] = 1 - scaler.fit_transform(df_resultado[['Costo de envío promedio']])

df_resultado['Índice de rendimiento'] = (
    df_resultado['Ventas'] * 0.5 +
    df_resultado['Calificaciones'] * 0.3 +
    df_resultado['Costo de envío promedio'] * 0.2
)

df_final = pd.concat([df_analisis, df_resultado['Índice de rendimiento']], axis=1)
peor_tienda = df_final.sort_values('Índice de rendimiento', ascending=True).iloc[0]

df_final['Ventas_Millones'] = df_final['Ventas'] / 1e6
peor_tienda['Ventas_Millones'] = peor_tienda['Ventas'] / 1e6

fig, ax = plt.subplots(figsize=(10, 6))
colormap = plt.colormaps['RdYlGn']
colores = colormap(df_final['Índice de rendimiento'])

scatter_plot = ax.scatter(
    df_final['Ventas_Millones'],
    df_final['Calificaciones'],
    s=2500,
    c=colores,
)

ax.scatter(
    peor_tienda['Ventas_Millones'],
    peor_tienda['Calificaciones'],
    s=2500,
    color='red',
)

sugerencia_final = ax.scatter(
    -100, -100,
    s=100,
    color='red',
    linewidth=1,
    label=f"La tienda con peor rendimiento es la {peor_tienda['Tienda']}"
)

for i, tienda in enumerate(df_final['Tienda']):
    ax.annotate(
        tienda,
        (df_final['Ventas_Millones'][i], df_final['Calificaciones'][i]),
        textcoords="offset points",
        xytext=(0, 0),
        ha='center',
        fontsize=9,
        fontweight='bold'
    )

ax.set_title('Analisís de rendimiento por tienda', fontsize=16, fontweight='bold')
ax.set_xlabel('Ventas Totales (Millones de $)', fontsize=12)
ax.set_ylabel('Calificación Promedio de Clientes', fontsize=12)

ax.set_xlim(df_final['Ventas_Millones'].min() - 20, df_final['Ventas_Millones'].max() + 20)
ax.set_ylim(df_final['Calificaciones'].min() - 0.05, df_final['Calificaciones'].max() + 0.05)


ax.grid(True, linestyle='--', alpha=0.7)
ax.legend(loc="lower left", frameon=True)

sm = ScalarMappable(cmap=colormap)
sm.set_array(df_final['Índice de rendimiento'])
cbar = plt.colorbar(sm, ax=ax)
cbar.set_label('Índice de Rendimiento (0-1)', rotation=270, labelpad=15)

plt.show()
```
